## How to fill in a cursor position table
A file object remembers **how far through the file it has read**. That position is called the **cursor**. It is invisible, so we write it down.

Suppose `notes.txt` contains exactly `Hello\nWorld\n` (12 characters, counting each newline as one).

**Worked example**

```python
f = open("notes.txt", "r")
print(repr(f.read(4)))
print(repr(f.readline()))
f.close()
```

| Step | Operation | What it returns | Cursor position after |
|---|---|---|---|
| 0 | file just opened | - | 0 |
| 1 | `f.read(4)` | `'Hell'` | 4 |
| 2 | `f.readline()` | `'o\n'` | 6 |

Step 2 is the important one. `readline()` did not return the whole first line, because the cursor was already at position 4. It returned only what was **left** of that line.

You can check any cursor position yourself with **`f.tell()`**.
