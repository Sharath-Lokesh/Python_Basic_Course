## How to fill in a name lookup table
When Python meets a name, it searches four places in a fixed order, called LEGB: Local, then Enclosing, then Global, then Built-in. It stops at the first place it finds the name.

**Worked example 1 - a name lookup table**

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        print(x)
    inner()

outer()
```

When `inner` runs and meets the name `x`, work through LEGB in order:

| Step | Where Python looks | Is `x` there? | Decision |
|---|---|---|---|
| 1 | Local - inside `inner` | No | keep searching |
| 2 | Enclosing - inside `outer` | Yes, "enclosing" | STOP - use this |
| 3 | Global - top level | Yes, "global" | never reached |
| 4 | Built-in | - | never reached |

So it prints `enclosing`. The global `x` still exists, but Python never got that far.

## How to trace a recursive call
A recursive function calls itself. To follow it, write the calls going DOWN until you hit the base case, then work the answers back UP.

**Worked example 2 - a recursion trace**

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(4))
```

Going down, each call has to wait for the one below it:

| Call | What it is waiting for | Value returned |
|---|---|---|
| factorial(4) | 4 * factorial(3) | 4 * 6 = 24 |
| factorial(3) | 3 * factorial(2) | 3 * 2 = 6 |
| factorial(2) | 2 * factorial(1) | 2 * 1 = 2 |
| factorial(1) | base case, waits for nothing | 1 |

Fill the "waiting for" column downwards first, then fill the "value returned" column upwards from the bottom. The answer is 24.

Q15. [Core] Write a decorator called `announce` that prints "Starting" before the function runs and "Done" after it runs, and still returns the function's result correctly. Test it on a function that adds two numbers.

````python
def announce(func):
    def wrapper():
        print("Starting")
        result = func()
        print("Done")
        return result
    return wrapper

@announce
def add():
    return 2 + 3

print(add())
````

Q16. [Stretch] Write a decorator that works with ANY function regardless of how many arguments it takes, using `*args` and `**kwargs`. Test it on two functions that take different numbers of arguments.

````python
def flexible(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@flexible
def add(a, b):
    return a + b

@flexible
def greet(name):
    return "Hello " + name

print(add(2, 3))
print(greet("Sam"))

print ("It works with both functions because *args and **kwargs can handle different arguments.")
````

Q17. [Stretch] Write a decorator `double_result` that doubles whatever number the decorated function returns. Test it on a function that returns the area of a circle.

````python
def double_result(func):
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs) * 2
    return wrapper

@double_result
def area(radius):
    return 3.14 * radius * radius

print(area(2))

print ("The normal area is 12.56, so it prints: 25.12")
````

````python
def log_mission(func):
    def wrapper(*args, **kwargs):
        print("Mission starting")
        result = func(*args, **kwargs)
        print("Mission finished")
        return result
    return wrapper


@log_mission
def launch(rocket, *crew, **details):
    print(f"Launching {rocket}")
    print(f"Crew: {crew}")
    print(f"Details: {details}")


launch("Falcon", "Alice", "Bob", destination="Mars", fuel="Full")
````
