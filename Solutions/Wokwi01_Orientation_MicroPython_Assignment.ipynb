Q3. [Stretch] Star Brightness Reporter. Store a dictionary of star names and their magnitudes. Loop through it and print each star with its magnitude, then blink the LED once for a bright star and twice for a faint one. Remember the astronomy rule: a smaller magnitude means a brighter star.

from machine import Pin
import time

led = Pin(25, Pin.OUT)

stars = {
    "Sirius": -1.46,
    "Vega": 0.03,
    "Polaris": 1.98,
    "Betelgeuse": 0.42
}

for star in stars:
    magnitude = stars[star]
    print(star, "magnitude:", magnitude)

    if magnitude < 1:
        led.value(1)
        time.sleep(0.3)
        led.value(0)
        time.sleep(0.3)
    else:
        for i in range(2):
            led.value(1)
            time.sleep(0.3)
            led.value(0)
            time.sleep(0.3)
