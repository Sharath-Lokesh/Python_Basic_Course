Q22. [Research] Write a code to print all the *prime numbers* until n where **2 < n < 200**. 
- Let n be an integer user input
- check if the value of n is within the boundary (2,200); if any other value throw an error

```
example:
if n = 10; Expected output: 2, 3, 5, 7
if n = 1; Expected output: ERROR! n must be with in 2 to 200 range
```

````python
n = int(input("Enter n: "))

if n <= 2 or n >= 200:
    print("ERROR! n must be within 2 to 200 range")
else:
    for num in range(2, n + 1):
        is_prime = True

        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            print(num)
````
