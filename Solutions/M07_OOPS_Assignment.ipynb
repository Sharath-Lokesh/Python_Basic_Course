## Tool 1: the object state table
Objects hold data that changes over time, and it is easy to lose track of which object holds what. An object state table has one row per step and one column per property.

**Worked example 1**

```python
class Rocket:
    fleet_size = 0

    def __init__(self, name, fuel):
        self.name = name
        self.fuel = fuel
        Rocket.fleet_size += 1

r1 = Rocket("Ares", 80)
r2 = Rocket("Orion", 50)
r1.fuel = 100
```

| Step | Line that ran | r1.name | r1.fuel | r2.name | r2.fuel | Rocket.fleet_size |
|---|---|---|---|---|---|---|
| 0 | class defined | - | - | - | - | 0 |
| 1 | `r1 = Rocket("Ares", 80)` | Ares | 80 | - | - | 1 |
| 2 | `r2 = Rocket("Orion", 50)` | Ares | 80 | Orion | 50 | 2 |
| 3 | `r1.fuel = 100` | Ares | 100 | Orion | 50 | 2 |

Notice that `fleet_size` has only ONE column. It belongs to the class and is shared, while each object has its own `name` and `fuel`.

## Tool 2: the attribute lookup table
When you write `obj.something`, Python searches for `something` in a fixed order, and stops at the first place it finds it. This is the object version of the LEGB rule you learned for scope.

**The search order:** the object itself, then its class, then the parent class, then the grandparent, and so on. If it is found nowhere, Python raises an `AttributeError`.

**Worked example 2**

```python
class Spacecraft:
    engines = 1

class Probe(Spacecraft):
    target = "Jupiter"

p = Probe()
p.name = "Juno"
print(p.engines)
```

| Step | Where Python looks | Is `engines` there? | Decision |
|---|---|---|---|
| 1 | the object `p` itself | No, it only has `name` | keep searching |
| 2 | its class `Probe` | No, it only has `target` | keep searching |
| 3 | the parent class `Spacecraft` | Yes, 1 | STOP - use this |

So it prints 1.

````python
class StarCatalogue:
    def __init__(self):
        self.stars = []

    def add(self, name, magnitude):
        self.stars.append((name, magnitude))

    def save(self, filename):
        with open(filename, "w") as file:
            for name, magnitude in self.stars:
                file.write(name + "," + str(magnitude) + "\n")

    def load(self, filename):
        with open(filename, "r") as file:
            for line in file:
                name, magnitude = line.strip().split(",")
                self.stars.append((name, float(magnitude)))

    def __len__(self):
        return len(self.stars)

    def __str__(self):
        return str(self.stars)


stars = StarCatalogue()
stars.add("Sun", -27)
stars.add("Sirius", -1.46)

stars.save("stars.txt")

new_stars = StarCatalogue()
new_stars.load("stars.txt")

print(new_stars)
````

````python
class Spacecraft:
    def __init__(self, name, fuel):
        self.name = name
        self.__fuel = fuel

    @property
    def fuel(self):
        return self.__fuel

    def __lt__(self, other):
        return self.fuel < other.fuel

    def __str__(self):
        return self.name


class Probe(Spacecraft):
    pass


class Shuttle(Spacecraft):
    pass


class Fleet:
    def __init__(self):
        self.ships = []

    def add(self, ship):
        self.ships.append(ship)

    def __len__(self):
        return len(self.ships)

    def __contains__(self, ship):
        return ship in self.ships


fleet = Fleet()
fleet.add(Probe("Probe", 100))
fleet.add(Shuttle("Shuttle", 500))

print(sorted(fleet.ships))
print(len(fleet))
````

Q8. [Stretch] Event System. Build a class `Button` that stores a list of callback functions, a method `on_press(func)` to register one, and a method `press()` that calls all of them. Register two normal functions and one bound method. Explain how this connects to your calculator's `command=` and to hardware interrupts.

````python
class Button:
    def __init__(self):
        self.callbacks = []

    def on_press(self, func):
        self.callbacks.append(func)

    def press(self):
        for func in self.callbacks:
            func()


def hello():
    print("Hello")


def bye():
    print("Bye")


class Robot:
    def move(self):
        print("Move")


robot = Robot()

button = Button()
button.on_press(hello)
button.on_press(bye)
button.on_press(robot.move)

button.press()


print ("This is like tkinter's command= because a function runs when the button is pressed. It is also similar to a hardware interrupt because an event causes a function to run.")
````
