#### Section A: You Have Been Using Abstraction All Along

Think about riding in a car. To make it go, the driver presses one pedal. To stop, another. To turn, the wheel.

The driver does **not** need to know how fuel is injected, how the gearbox chooses a gear, or how the brake pads squeeze the discs. All that machinery is real, and it is complicated, but it is **hidden behind three simple controls**.

That is the whole idea you are about to learn. It has a name: **abstraction**.

> **Abstraction means hiding the complicated details and offering a simple way to use something.**

You show the **what** (press this pedal to go) and hide the **how** (the entire engine).

#### Section C: The Interface: What the User Needs to Know

The set of things a class offers to the outside world is called its **interface**. Think of it as the dashboard of a car, or the buttons on a TV remote.

- The **interface** is the public method names, what you pass them, and what you get back.
- The **implementation** is everything inside those methods. It is nobody else's business.

A useful test: **could you use this class correctly if you were only shown the method names and one line describing each?** If yes, the abstraction is good.

Here is a class described only by its interface. The bodies are hidden from you on purpose.

```
class Telescope:
    Telescope(name)            creates a telescope with a name
    point_at(target)           aims the telescope at a named target
    current_target()           returns the target it is aimed at now
    history_count()            returns how many targets it has been aimed at
```

Q1. [Core] Using only the interface description above, write the code that creates a telescope called "Hubble Jr", points it at Mars, then at Andromeda, then prints the current target and the number of targets in its history. You have never seen inside this class. Can you still write the code?

Q2. [Core] Before running anything, predict what your code will print. Then run this complete version and check.
```python
class Telescope:
    def __init__(self, name):
        self.name = name
        self.__target = None
        self.__log = []

    def point_at(self, target):
        self.__target = target
        self.__log.append(target)

    def current_target(self):
        return self.__target

    def history_count(self):
        return len(self.__log)

t = Telescope("Hubble Jr")
t.point_at("Mars")
t.point_at("Andromeda")
print(t.current_target())
print(t.history_count())
```

#### Section D: The Payoff: Changing the Inside Without Breaking the Outside

This section shows the biggest reason abstraction is worth the effort.

If the interface stays the same, you can **completely rewrite the inside of a class**, and every program that uses it keeps working without a single change.

Below are two versions of the same class. Look carefully: the insides are completely different, but the method names and what they return are identical.

Q1. [Core] Predict what this prints.
```python
class ReadingLog:
    def __init__(self):
        self.readings = []

    def add(self, value):
        self.readings.append(value)

    def average(self):
        total = 0
        for r in self.readings:
            total += r
        return total / len(self.readings)

log = ReadingLog()
log.add(10)
log.add(20)
log.add(30)
print(log.average())
```

Q2. [Core] Now predict what THIS prints. The inside is completely different: it does not store the readings at all, only a running total and a count.
```python
class ReadingLog:
    def __init__(self):
        self.total = 0
        self.count = 0

    def add(self, value):
        self.total += value
        self.count += 1

    def average(self):
        return self.total / self.count

log = ReadingLog()
log.add(10)
log.add(20)
log.add(30)
print(log.average())
```

#### Section E: Abstraction and Encapsulation Are Not the Same Thing

These two are easy to mix up, because they work together and both involve hiding. Here is the difference.

| | **Encapsulation** | **Abstraction** |
|---|---|---|
| What it hides | the **data** | the **complexity** |
| Why | to stop the data being damaged | to make the thing simple to use |
| How, in Python | `__private` properties, getters and setters | well-chosen public methods |
| The question it answers | "who is allowed to touch this value?" | "what does someone need to know to use this?" |

A rocket with a private `__fuel` that can never go negative is **encapsulation**.

A rocket with a single `can_launch()` method that hides four separate checks is **abstraction**.

Most good classes use both at once.

#### Section F: Layers of Abstraction

Abstraction is not only for the outside of a class. You can use it **inside** a class too, by splitting one complicated method into several simple ones.

By convention, a method meant only for use inside the class gets a single underscore at the front, exactly like a protected property.

```python
class Rocket:
    def __init__(self, name, fuel, crew):
        self.name = name
        self.fuel = fuel
        self.crew = crew

    def _check_fuel(self):
        print("  checking fuel")
        return self.fuel >= 100

    def _check_crew(self):
        print("  checking crew")
        return len(self.crew) >= 3

    def _ignite(self):
        print("  ignition")

    def launch(self):
        print("launch sequence for", self.name)
        if self._check_fuel() and self._check_crew():
            self._ignite()
            print(self.name, "has launched")
        else:
            print("launch aborted")
```

The outside world sees one method: `launch()`. Inside, the work is split into small pieces that each do one job.

#### Section G: A Shared Interface: Promising Every Child Has the Method

Abstraction becomes really powerful when several different classes agree to offer the **same interface**.

If every spacecraft promises to have a `launch()` method, then any code can launch any spacecraft without caring which kind it is. You saw this in Module 7 and it was called polymorphism. Abstraction is the **design decision** that makes polymorphism possible: you decide on the shared interface first.

One way to record that promise is to put the method in the parent class with a message reminding the programmer to replace it.

```python
class Spacecraft:
    def launch(self):
        print("Every spacecraft must define its own launch method")
```

#### Section H: Formal Abstract Base Classes

Python has a proper tool for the promise you made in Section G. It lives in a module called **`abc`**, which stands for **a**bstract **b**ase **c**lass.

- You import two things: `ABC` and `abstractmethod`.
- Your parent class inherits from `ABC`.
- You mark each required method with the `@abstractmethod` decorator, and leave its body as `pass`.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
```

Two things now happen automatically:

1. You cannot create an object of `Shape` itself. It is only a blueprint for other blueprints.
2. Any child class that forgets to write `area()` cannot be used either, and Python tells you **immediately** rather than later.

This is the same promise as Section G, but now Python enforces it for you.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    def describe(self):
        print("My area is", self.area())

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return round(3.14159 * self.r ** 2, 2)

for sh in (Square(4), Circle(2)):
    sh.describe()
```
