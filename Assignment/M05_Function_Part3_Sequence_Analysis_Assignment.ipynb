# Python Assignment: Function Sequence Analysis - How Control Flows

This assignment is about ONE question, asked over and over in different situations:

> **Who is in control right now, and where does control go next?**

A normal call, a `return`, a nested call, recursion, a decorator, a generator, a callback and a hardware interrupt are all different answers to that same question. By the end you should be able to take any program built from the functions you have learned and write down the exact order in which every line runs.

## The call stack, in plain language
When your program calls a function, the line that made the call has to WAIT. Python writes down where it was, jumps into the function, and comes back to that exact spot afterwards.

Think of a stack of plates. Every time a function is called, a new plate goes on top. Python can only ever work on the top plate. When a function returns, its plate is taken off and Python carries on with the plate underneath. That pile of plates is called the **call stack**.

At the very bottom is the plate we will call `main` - the top level of your program, outside any function.

## How to fill in an execution order table
Number the steps in the order they actually happen, not the order they are written on the page.

**Worked example**

```python
def greet(name):
    print("inside greet")
    return "Hello " + name

print("start")
msg = greet("Ada")
print(msg)
print("end")
```

| Step | What runs | Output | Call stack after this step |
|---|---|---|---|
| 1 | `print("start")` | start | main |
| 2 | `greet("Ada")` is called, control jumps INTO greet | - | main > greet |
| 3 | `print("inside greet")` | inside greet | main > greet |
| 4 | `return "Hello Ada"` - control jumps BACK | - | main |
| 5 | `msg` now holds the returned value, `print(msg)` runs | Hello Ada | main |
| 6 | `print("end")` | end | main |

Notice that step 4 is the important one. The `return` does two separate jobs at once: it hands a value back, AND it moves control back to the line that made the call.
