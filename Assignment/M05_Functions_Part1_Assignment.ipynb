## How to fill in a trace table
A trace table has one row for each pass of the loop and one column for each variable that changes. You fill it in one row at a time, in order. You may type into the table in the markdown cell, or draw it on paper and photograph it.

**Worked example 1 - a `for` loop**

```python
def add_all(numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(add_all([5, 2, 4]))
```

Before the loop starts, `total` is 0. Now take it one pass at a time:

| Pass | num | total BEFORE | total AFTER |
|---|---|---|---|
| 1 | 5 | 0 | 5 |
| 2 | 2 | 5 | 7 |
| 3 | 4 | 7 | 11 |

The list has run out, so the loop stops and the function returns 11.

**Worked example 2 - a `while` loop**

```python
i = 1
total = 0
while i <= 3:
    total += i
    i += 1
print(total)
```

With a `while` loop, always check the condition before each pass:

| Pass | i at start | Is i <= 3? | total AFTER | i AFTER |
|---|---|---|---|---|
| 1 | 1 | Yes | 1 | 2 |
| 2 | 2 | Yes | 3 | 3 |
| 3 | 3 | Yes | 6 | 4 |
| 4 | 4 | No - loop stops | 6 | 4 |

So the program prints 6. Notice that the last row is where the condition FAILS, which is what ends the loop.
