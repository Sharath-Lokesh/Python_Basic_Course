Q1. [Core] Read this measuring code carefully, then predict what one reading would print if the slider is set to about 100 cm:
```python
from machine import Pin, time_pulse_us
import time

trigger = Pin(17, Pin.OUT)
echo = Pin(16, Pin.IN)

trigger.value(0)
time.sleep_us(5)
trigger.value(1)
time.sleep_us(10)
trigger.value(0)

duration = time_pulse_us(echo, 1, 30000)
distance = duration * 0.0343 / 2
print(distance)
```
