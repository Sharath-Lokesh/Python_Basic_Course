# Module 6: File Handling in Python

### File Handling
- So far, every value our programs created disappeared the moment the program finished.
- A **file** lets us store data on the computer's disk, so it is still there the next time the program runs.
- This is the difference between **temporary memory** (variables) and **permanent storage** (files).
- Python can **create**, **read**, **update** and **delete** files.
- Two kinds of files:
  - **Text files** - hold readable characters, for example `.txt` and `.csv`
  - **Binary files** - hold raw bytes, for example images and sound files

#### The open() Function
- The key function for working with files is the **`open()`** function.
- `open()` takes two parameters: the **filename** and the **mode**.
- **Syntax**
```
f = open(filename, mode)
```
- It returns a **file object**, which we store in a variable (commonly called `f`).

##### The four main modes
| Mode | Name | What it does |
|---|---|---|
| `"r"` | Read | Default value. Opens a file for reading. **Error** if the file does not exist |
| `"a"` | Append | Opens a file for adding to the end. **Creates** the file if it does not exist |
| `"w"` | Write | Opens a file for writing. **Creates** the file if it does not exist, and **erases** it if it does |
| `"x"` | Create | Creates the file. **Error** if the file already exists |

##### Text or binary
| Mode | Name | What it does |
|---|---|---|
| `"t"` | Text | Default value. Text mode |
| `"b"` | Binary | Binary mode, for example images |

- The mode letters are combined, so `"rt"` means read + text.
- Because `"r"` and `"t"` are the **defaults**, these two lines mean exactly the same thing:
```
f = open("demofile.txt")
f = open("demofile.txt", "rt")
```
- ***Note:*** For mode `"r"`, the file must already exist, or Python raises a `FileNotFoundError`.

#### Preparing a File to Work With
Before we can read a file, one has to exist. The next cell creates a small demo file that the rest of these notes will use.

````python
# create a file to use in the examples below
f = open("demofile.txt", "w")
f.write("Hello! Welcome to demofile.txt\n")
f.write("This file is for testing purposes.\n")
f.write("Good Luck!\n")
f.close()

print("demofile.txt has been created")
````

### Reading Files
- Once a file is open in read mode, the file object gives us several ways to get the text out.

#### The read() Method
- **`read()`** returns the **whole file** as one single string.

````python
f = open("demofile.txt", "r")
print(f.read())
f.close()
````

#### Reading Only Part of the File
- By passing a number to `read()`, we return only that many **characters**.

````python
f = open("demofile.txt", "r")
print(f.read(6))    # only the first 5 characters
f.close()
````

#### The readline() Method
- **`readline()`** returns **one line** at a time.
- Calling it again returns the **next** line, and so on.

````python
f = open("demofile.txt", "r")
print(f.readline())
f.close()
````

````python
# calling readline() twice reads the first two lines
f = open("demofile.txt", "r")
print(f.readline())
print(f.readline())
f.close()
````

#### The File Cursor
- A file object keeps track of **where it has read up to**. This position is called the **cursor** or **file pointer**.
- Every read moves the cursor forward. It never goes back on its own.
- This explains a common surprise: reading a file a second time returns an **empty string**, because the cursor is already at the end.
- ***Note:*** This is the same idea as a generator remembering where it paused. The file object remembers its position between calls.

````python
f = open("demofile.txt", "r")
first = f.read()
second = f.read()       # the cursor is already at the end of the file
print("First read got", len(first), "characters")
print("Second read got", len(second), "characters")
f.close()
````

- To go back to the beginning, use **`seek(0)`**.

````python
f = open("demofile.txt", "r")
print("Before seek:", len(f.read()), "characters")
f.seek(0)               # move the cursor back to the start
print("After seek: ", len(f.read()), "characters")
f.close()
````

#### The readlines() Method
- **`readlines()`** returns **every line as a list of strings**.
- Each string still ends with the newline character `\n`.

````python
f = open("demofile.txt", "r")
lines = f.readlines()
f.close()

print(lines)
print("Number of lines:", len(lines))
````

#### Looping Through a File
- A file object can be used directly in a **`for` loop**, which reads it one line at a time.
- This is the preferred way to read a large file, because the whole file never has to fit in memory at once.

````python
f = open("demofile.txt", "r")
for line in f:
    print(line)
f.close()
````

- Notice the **blank lines** in the output above. Each line already ends with `\n`, and `print()` adds another one.
- Use **`strip()`** to remove the newline, or `end=""` in `print()`.

````python
f = open("demofile.txt", "r")
for line in f:
    print(line.strip())     # strip() removes the newline at the end
f.close()
````

#### Closing a File
- **`close()`** tells the operating system that we have finished with the file.
- ***Note:*** Always close your files. Because of **buffering**, text you have written may not actually appear in the file until it is closed.

### The with Statement
- It is easy to forget `close()`, and if an error happens the file may never be closed at all.
- The **`with`** statement solves this: it closes the file **automatically**, even if something goes wrong.
- **Syntax**
```
with open(filename, mode) as f:
    # work with the file here
# the file is closed automatically here
```
- This is the **recommended** way to work with files.

````python
with open("demofile.txt", "r") as f: #f = open(filename, mode)
    content = f.read()

print(content)
print("Is the file closed now?", f.closed)
````

### Writing to Files
- To write to a file, open it with one of the writing modes.
- Two modes are used most often:
  - **`"a"` Append** - adds to the **end** of the file, keeping what is already there
  - **`"w"` Write** - **erases everything** in the file first, then writes
- The method used is **`write()`**.

#### Appending with "a"
- Appending keeps the existing content and adds to the end.

````python
# append a line to the existing file
with open("demofile.txt", "a") as f:
    f.write("This line was appended.\n")

# open and read the file after the appending
with open("demofile.txt", "r") as f:
    print(f.read())
````

#### Overwriting with "w"
- ***Note:*** The `"w"` mode **deletes the entire contents** of the file before writing. This cannot be undone, so use it carefully.

````python
# the "w" mode will overwrite the entire file
with open("demofile.txt", "w") as f:
    f.write("All the old content is gone.\n")

with open("demofile.txt", "r") as f:
    print(f.read())
````

#### Writing Several Lines
- **`write()`** does **not** add a newline by itself. If we want separate lines, we must include `\n`.
- **`writelines()`** writes every string in a list, again without adding newlines.

````python
planets = ["Mercury\n", "Venus\n", "Earth\n", "Mars\n"]

with open("planets.txt", "w") as f:
    f.writelines(planets)

with open("planets.txt", "r") as f:
    print(f.read())
````

- A `for` loop gives more control, and lets us add the newline ourselves.

````python
moons = {"Earth": 1, "Mars": 2, "Jupiter": 95}

with open("moons.txt", "w") as f:
    for planet, count in moons.items():
        f.write(f"{planet} has {count} moon(s)\n")

with open("moons.txt", "r") as f:
    print(f.read())
````

### Creating New Files
Three modes will create a file that does not exist yet:

| Mode | Behaviour |
|---|---|
| `"x"` | Creates the file, and raises an **error if it already exists** |
| `"a"` | Creates the file if it does not exist |
| `"w"` | Creates the file if it does not exist |

- Use `"x"` when you want to be **certain** you are not overwriting something.

````python
# "x" creates a brand new file
with open("mission_log.txt", "x") as f:
    f.write("Mission log created.\n")

print("mission_log.txt created successfully")
````

````python
# running "x" again on the same file raises an error
try:
    f = open("mission_log.txt", "x")
except FileExistsError:
    print("FileExistsError: that file already exists")
````

### Deleting Files
- Deleting is **not** done with `open()`. It is handled by the **`os`** module, which talks to the operating system.
- Import it first with `import os`.

#### Deleting a File

````python
import os

os.remove("planets.txt")
print("planets.txt has been deleted")
````

#### Checking if a File Exists First
- Trying to delete a file that is not there raises a `FileNotFoundError`.
- **`os.path.exists()`** returns `True` or `False`, so we can check before deleting.

````python
import os

if os.path.exists("moons.txt"):
    os.remove("moons.txt")
    print("moons.txt has been deleted")
else:
    print("The file does not exist")
````

````python
# running the same code again now takes the else branch
import os

if os.path.exists("moons.txt"):
    os.remove("moons.txt")
    print("moons.txt has been deleted")
else:
    print("The file does not exist")
````

#### Deleting a Folder
- **`os.rmdir()`** removes a folder.
- ***Note:*** You can only remove **empty** folders.

````python
import os

os.mkdir("myfolder")            # create a folder
print("Folder exists:", os.path.exists("myfolder"))

os.rmdir("myfolder")            # remove the folder
print("Folder exists:", os.path.exists("myfolder"))
````

### File Paths
- So far the files have been created in the **same folder** as the notebook. That is called a **relative path**.
- An **absolute path** gives the full location, starting from the top of the drive.
- **Examples**
```
f = open("demofile.txt")                  # same folder (relative)
f = open("data/demofile.txt")             # inside a subfolder called data
f = open("C:/Users/Student/demofile.txt") # full location (absolute)
```
- ***Note:*** Use forward slashes `/` in paths. A backslash `\` starts an escape character in Python, so `"C:\new"` would be read as a newline.

````python
import os

# where is Python currently looking for files?
print(os.getcwd())
````

### Common File Errors
| Error | Usual cause |
|---|---|
| `FileNotFoundError` | Opening a file with `"r"` when the file does not exist, or the path is wrong |
| `FileExistsError` | Using `"x"` on a file that already exists |
| `PermissionError` | The file is protected, or is open in another program |
| `UnicodeDecodeError` | Reading a binary file such as an image in text mode |

- These errors can be handled with `try` and `except`.

````python
try:
    with open("a_file_that_does_not_exist.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("FileNotFoundError: that file could not be found")
````

### Putting It All Together
A small example that uses the ideas from this module together with loops, dictionaries, functions and f-strings from earlier modules.

````python
# write a list of sensor readings to a file, one per line
def save_readings(filename, readings):
    number = 1
    with open(filename, "w") as f:
        for value in readings:
            f.write(f"Reading {number}: {value} cm\n")
            number += 1


# read the file back and return the lines as a list
def load_readings(filename):
    lines = []
    with open(filename, "r") as f:
        for line in f:
            lines.append(line.strip())
    return lines


distances = [120, 85, 40, 15]

save_readings("sensor_log.txt", distances)
print("Saved.\n")

for line in load_readings("sensor_log.txt"):
    print(line)
````

### Summary
- `open(filename, mode)` opens a file and returns a **file object**.
- Modes: **`"r"`** read, **`"a"`** append, **`"w"`** write (erases), **`"x"`** create. Add **`"b"`** for binary.
- Reading: **`read()`**, **`read(n)`**, **`readline()`**, **`readlines()`**, or a **`for` loop** over the file.
- The file object keeps a **cursor** showing where it has read up to. **`seek(0)`** returns to the start.
- Writing: **`write()`** and **`writelines()`**. Neither adds a newline, so include `\n` yourself.
- Always close a file. The **`with`** statement does this automatically and is the recommended approach.
- Deleting uses the **`os`** module: **`os.remove()`** for files and **`os.rmdir()`** for empty folders.
- Check with **`os.path.exists()`** before deleting, to avoid an error.
