## Classes, Objects, __init__, self, Properties and Methods

### Why Object-Oriented Programming?
Think back to the calculator you built. It had a variable called `expression`, and every function had to start with the line `global expression` just to reach it.

- The **data** (`expression`) lived in one place.
- The **functions** that worked on it (`press`, `clear`, `equal`) lived somewhere else.
- Nothing tied them together, so we had to use `global` to bridge the gap.

That is the problem **Object-Oriented Programming** solves. It lets you put the data and the functions that work on that data **inside the same box**. The box is called an **object**.

If you wanted two calculators on screen at once, the old design could not do it. There was only one `expression`. With objects, you simply make two of them, and each keeps its own.

### Classes and Objects
- Python is an **object-oriented programming language**. Almost everything in Python is an object, with its properties and methods.
- A **class** is a **blueprint**. It describes what something should have and what it should be able to do.
- An **object** is a **real thing built from that blueprint**.

A useful picture:
- The class is the **architect's drawing of a house**.
- Each object is an **actual house built from that drawing**.
- One drawing, many houses. Each house has its own address and its own furniture.

You have used objects all along without knowing it. A string has methods like `upper()`. A list has methods like `append()`. Those come from the `str` class and the `list` class.

#### Creating a Class
- Use the **`class`** keyword. By convention, class names start with a **capital letter**.
- **Syntax**
```
class ClassName:
    # properties and methods go here
```

````python
# a very simple class with one property
class MyClass:
    x = 5

print(MyClass)
````

#### Creating an Object
- To build an object from the class, write the class name followed by **parentheses**.
- This is called **creating an instance** of the class. The object is an **instance**.

````python
p1 = MyClass()
print(p1.x)
````

#### Creating Several Objects
- One class can build as many objects as you like.

````python
p1 = MyClass()
p2 = MyClass()
p3 = MyClass()

print(p1.x)
print(p2.x)
print(p3.x)
````

### The __init__() Method
- The simple class above gives every object exactly the same `x`. That is not very useful.
- We usually want each object to start with **its own values**.
- The **`__init__()`** method runs **automatically** every time a new object is created. Its job is to set up the starting values.
- It is often called the **constructor**.
- ***Note:*** The name has **two underscores** on each side. Python looks for that exact name.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("John", 36)

print(p1.name)
print(p1.age)
````

- Because `__init__()` runs automatically, the values in the parentheses when you create the object are passed straight to it.
- Each object gets **its own** `name` and `age`.

````python
p1 = Person("John", 36)
p2 = Person("Ada", 12)

print(p1.name, p1.age)
print(p2.name, p2.age)
````

### The self Parameter
This is the part that confuses everybody at first, so read it slowly.

- `self` is simply the **first parameter of every method**.
- It refers to **the particular object the method was called on**.
- You do **not** pass it yourself. Python fills it in automatically.

When you write `p1.greet()`, Python secretly turns it into `Person.greet(p1)`. The object goes in as the first argument. That is all `self` is.

- ***Note:*** The name `self` is only a **convention**. You may call it anything, as long as it is the **first** parameter. The next example proves it, but you should always use `self` in real code.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("John")
p2 = Person("Ada")

p1.greet()      # self becomes p1
p2.greet()      # self becomes p2
````

````python
# proof that the name "self" is just a convention
class Person:
    def __init__(mysillyobject, name):
        mysillyobject.name = name

    def greet(abc):
        print("Hello, my name is " + abc.name)

p = Person("Ada")
p.greet()
````

#### What Happens if You Forget self
- If a method has no first parameter, Python still passes the object, and the call fails.
- This is one of the most common beginner errors, so learn to recognise it.

````python
class Person:
    def greet():          # no self here
        print("Hello")

p = Person()

try:
    p.greet()
except TypeError as e:
    print("TypeError:", e)
````

### Methods
- A **method** is simply a **function that lives inside a class**.
- Everything you learned about functions still applies: parameters, default values, `return`, `*args` and `**kwargs`.
- The only new rule is that the first parameter is `self`.

````python
class Rocket:
    def __init__(self, name, fuel):
        self.name = name
        self.fuel = fuel

    def launch(self):
        if self.fuel >= 100:
            print(self.name, "has launched")
        else:
            print(self.name, "does not have enough fuel")

    def add_fuel(self, amount):
        self.fuel = self.fuel + amount
        return self.fuel

r1 = Rocket("Falcon", 80)
r1.launch()
print("Fuel is now", r1.add_fuel(50))
r1.launch()
````

### Instance Properties and Class Properties
There are **two different kinds** of property, and mixing them up causes real bugs.

| Kind | Where it is created | Who owns it |
|---|---|---|
| **Instance property** | inside `__init__()`, using `self.` | each object has **its own** |
| **Class property** | directly inside the class body | **shared by every object** |

- ***Note:*** Some tutorials say each object gets its own copy of a class property. That is **not correct**, and the next two cells prove it.

````python
class MyClass:
    x = 5           # class property, written directly in the class body

p1 = MyClass()
p2 = MyClass()

MyClass.x = 99      # change it once, on the class

print(p1.x)
print(p2.x)         # both objects see the change, so it was never copied
````

- The danger appears when a class property is something changeable, such as a **list**.

````python
class Rocket:
    crew = []                      # class property - ONE list, shared by all

    def add(self, name):
        self.crew.append(name)

r1 = Rocket()
r2 = Rocket()

r1.add("Ada")

print("r1.crew:", r1.crew)
print("r2.crew:", r2.crew)         # r2 never added anyone
````

- The fix is to create the list **inside `__init__()`**, so each object builds its own.

````python
class Rocket:
    def __init__(self):
        self.crew = []             # instance property - a NEW list per object

    def add(self, name):
        self.crew.append(name)

r1 = Rocket()
r2 = Rocket()

r1.add("Ada")

print("r1.crew:", r1.crew)
print("r2.crew:", r2.crew)
````

- A class property is still useful when the value genuinely **belongs to the whole class**, such as a fixed constant.

````python
class Circle:
    pi = 3.14159                   # the same for every circle ever made

    def __init__(self, radius):
        self.radius = radius       # different for each circle

    def area(self):
        return Circle.pi * self.radius * self.radius

print(Circle(2).area())
print(Circle(5).area())
````

#### Assigning to an Object Creates an Instance Property
- If you assign to `p1.x`, Python does **not** change the class property. It creates a **new instance property** on that object only, which hides the class one.

````python
class MyClass:
    x = 5

p1 = MyClass()
p2 = MyClass()

p1.x = 100          # creates an instance property on p1 only

print("p1.x     :", p1.x)
print("p2.x     :", p2.x)
print("MyClass.x:", MyClass.x)
````

### Modifying and Deleting
#### Modify an Object Property

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("John", 36)
p1.age = 40
print(p1.age)
````

#### Delete an Object Property
- Use the **`del`** keyword.

````python
del p1.age

try:
    print(p1.age)
except AttributeError as e:
    print("AttributeError:", e)
````

#### Delete an Object
- `del` also removes the whole object.

````python
p1 = Person("John", 36)
del p1

try:
    print(p1)
except NameError as e:
    print("NameError:", e)
````

### The pass Statement
- A class definition **cannot be empty**. If you want to write the class now and fill it in later, use **`pass`**.

````python
class Person:
    pass

print(Person())
````

### Putting It Together
Here is the calculator idea from the start of these notes, rebuilt as a class. Notice that there is no `global` anywhere, and that two calculators can exist at the same time without interfering.

````python
class Calculator:
    def __init__(self):
        self.expression = ""       # each calculator gets its own

    def press(self, value):
        self.expression = self.expression + value

    def clear(self):
        self.expression = ""

    def show(self):
        print("Display:", self.expression)


calc1 = Calculator()
calc2 = Calculator()

calc1.press("1")
calc1.press("2")
calc1.press("+")
calc1.press("3")

calc2.press("9")

calc1.show()
calc2.show()

calc1.clear()
calc1.show()
calc2.show()
````

### Summary
- A **class** is a blueprint. An **object** is a thing built from that blueprint.
- Create a class with the **`class`** keyword, and an object by calling the class name with parentheses.
- **`__init__()`** runs automatically when an object is created, and sets up its starting values.
- **`self`** is the first parameter of every method and refers to the object the method was called on. Python passes it for you.
- A **method** is a function inside a class.
- **Instance properties** (`self.x` in `__init__`) belong to one object. **Class properties** (written in the class body) are **shared by all objects**.
- Use **`del`** to remove a property or a whole object, and **`pass`** for a class you have not written yet.
