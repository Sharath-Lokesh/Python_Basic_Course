## Inheritance and Polymorphism

### Inheritance
- **Inheritance** lets us define a class that takes over all the methods and properties of another class.
- The **parent class** (also called the base class) is the one being inherited from.
- The **child class** (also called the derived class) is the one that inherits.
- The point is to **avoid repeating yourself**. If two classes share most of their behaviour, write it once in the parent.

#### Creating a Parent Class
- Any class can be a parent. It is written exactly as normal.

````python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)

x = Person("John", "Doe")
x.printname()
````

#### Creating a Child Class
- Put the parent class name **inside the parentheses** of the child class.
- **Syntax**
```
class ChildName(ParentName):
    ...
```
- If the child needs nothing of its own yet, use `pass`. It still gets everything from the parent.

````python
class Student(Person):
    pass

s = Student("Mike", "Olsen")
s.printname()          # this method came from Person
````

#### Adding an __init__() to the Child
- Often the child needs extra setup of its own, so it defines its own `__init__()`.
- ***Note:*** As soon as the child defines `__init__()`, the parent's `__init__()` is **no longer called automatically**. You must call it yourself.

````python
class Student(Person):
    def __init__(self, fname, lname, year):
        Person.__init__(self, fname, lname)    # call the parent's setup
        self.graduationyear = year

s = Student("Mike", "Olsen", 2019)
s.printname()
print(s.graduationyear)
````

#### The super() Function
- **`super()`** refers to the parent class, without you having to name it.
- It is the preferred way, because if the parent's name ever changes, this code still works.
- ***Note:*** With `super()` you do **not** pass `self`. Python handles it.

````python
class Student(Person):
    def __init__(self, fname, lname, year):
        super().__init__(fname, lname)         # cleaner than naming Person
        self.graduationyear = year

    def welcome(self):
        print("Welcome", self.firstname, self.lastname,
              "to the class of", self.graduationyear)

s = Student("Mike", "Olsen", 2019)
s.printname()
s.welcome()
````

### Method Overriding
- A child class can **replace** a method it inherited, simply by defining a method with the **same name**.
- This is called **overriding**. Python always uses the version closest to the object.

````python
class Animal:
    def speak(self):
        print("Some generic sound")

class Dog(Animal):
    def speak(self):                 # overrides the parent version
        print("Woof")

class Cat(Animal):
    def speak(self):
        print("Meow")

Animal().speak()
Dog().speak()
Cat().speak()
````

- You can also **extend** the parent's method instead of replacing it completely, by calling `super()` inside the override.

````python
class Rocket:
    def status(self):
        print("Rocket systems nominal")

class CrewedRocket(Rocket):
    def status(self):
        super().status()             # do the parent's work first
        print("Crew life support nominal")

CrewedRocket().status()
````

### Multi-Level Inheritance
- A child class can itself be a parent. Python searches **upwards** until it finds the method.

````python
class Vehicle:
    def describe(self):
        print("I am a vehicle")

class Car(Vehicle):
    pass

class SportsCar(Car):
    pass

SportsCar().describe()               # found all the way up in Vehicle
````

#### Checking Relationships
- **`isinstance(object, Class)`** asks whether an object was built from that class, or from one of its children.
- **`issubclass(Child, Parent)`** asks whether one class inherits from another.

````python
class Animal: pass
class Dog(Animal): pass

d = Dog()

print(isinstance(d, Dog))
print(isinstance(d, Animal))         # a Dog IS an Animal
print(issubclass(Dog, Animal))
print(issubclass(Animal, Dog))       # but an Animal is not a Dog
````

### Polymorphism
- The word **polymorphism** means "many forms".
- In programming it means: **the same method name behaves differently depending on the object it is called on**.
- You have already used this. `len()` works on a string, a list and a dictionary, and does the right thing for each.

````python
print(len("Mars"))          # counts characters
print(len([1, 2, 3]))        # counts items
print(len({"a": 1}))         # counts keys
````

#### Polymorphism with Separate Classes
- The classes below are **not related at all**, yet each has a `move()` method.
- Because of that, one loop can drive all of them.

````python
class Car:
    def __init__(self, brand):
        self.brand = brand
    def move(self):
        print("Drive!")

class Boat:
    def __init__(self, brand):
        self.brand = brand
    def move(self):
        print("Sail!")

class Plane:
    def __init__(self, brand):
        self.brand = brand
    def move(self):
        print("Fly!")

for vehicle in (Car("Ford"), Boat("Ibiza"), Plane("Boeing")):
    print(vehicle.brand)
    vehicle.move()
````

- Notice what the loop did **not** need to do: it never checked what kind of object it had. No `if` statements, no `match`. It simply called `move()` and each object knew what to do.
- This is the real benefit of polymorphism. Adding a `Submarine` class later would need **no change at all** to the loop.

#### Polymorphism with Inheritance
- The same idea works when the classes share a parent. Each child overrides the method in its own way.

````python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
    def move(self):
        print("Move!")

class Car(Vehicle):
    pass                             # happy with the inherited version

class Boat(Vehicle):
    def move(self):
        print("Sail!")

class Plane(Vehicle):
    def move(self):
        print("Fly!")

for vehicle in (Car("Ford"), Boat("Ibiza"), Plane("Boeing")):
    print(vehicle.brand)
    vehicle.move()
````

### A Space Example

````python
class Spacecraft:
    def __init__(self, name):
        self.name = name

    def describe(self):
        print(self.name, "is a spacecraft")

class Orbiter(Spacecraft):
    def describe(self):
        print(self.name, "orbits a planet and takes measurements")

class Lander(Spacecraft):
    def describe(self):
        print(self.name, "touches down on the surface")

class Rover(Lander):
    def describe(self):
        super().describe()
        print(self.name, "then drives around and collects samples")

fleet = [Spacecraft("Voyager"), Orbiter("Juno"),
         Lander("Viking"), Rover("Perseverance")]

for craft in fleet:
    craft.describe()
    print()
````

### Summary
- **Inheritance** lets a child class take over the properties and methods of a parent class, so shared behaviour is written once.
- Write the parent name in the child's parentheses: `class Child(Parent):`.
- If the child defines its own `__init__()`, the parent's is no longer called automatically. Use **`super().__init__(...)`** to call it.
- **Overriding** means a child defines a method with the same name as the parent's, replacing it. Calling `super()` inside lets you extend rather than replace.
- Python searches **upwards** through the parents until it finds a method.
- **`isinstance()`** and **`issubclass()`** check these relationships.
- **Polymorphism** means the same method name behaves differently for different objects, which lets one loop handle many types without any `if` statements.
